```python
import numpy as np
import pandas as pd
import yfinance as yf

# ============================================
# 1. ASSET UNIVERSE (12 markets)
# ============================================
universe = {
    'bonds': ['TLT', 'IEF', 'BNDX', 'IGOV'],
    'currencies': ['UUP', 'FXE', 'FXY', 'FXB'],
    'equities': ['SPY', 'EWJ', 'EFA', 'EEM'],
    'commodities': ['GLD', 'DBC', 'USO', 'CORN']
}

# Flatten into one list
all_tickers = [t for sublist in universe.values() for t in sublist]
print(f"Universe: {len(all_tickers)} markets")
print(all_tickers)

# ============================================
# 2. DOWNLOAD DATA
# ============================================
print("\nDownloading data...")
data = yf.download(all_tickers, period='5y', progress=False, auto_adjust=True)['Close']
data = data.dropna(how='all')
print(f"Got {len(data)} days of data")
print(data.tail())

# ============================================
# 3. MOMENTUM SIGNAL (12-month return)
# ============================================
def momentum_signal(prices):
    return prices.pct_change(252).iloc[-1]

momentum = momentum_signal(data)
print("\n=== MOMENTUM SIGNALS (12-month return) ===")
print(momentum.sort_values(ascending=False).round(3))


# ============================================
# 4. VALUE SIGNAL (5-year z-score, inverse for mean reversion)
# ============================================
def value_signal(prices):
    mean = prices.rolling(252*3).mean().iloc[-1]
    std = prices.rolling(252*3).std().iloc[-1]
    zscore = (prices.iloc[-1] - mean) / std
    return -zscore  # Negative because high z-score = expensive = short

value = value_signal(data)
print("\n=== VALUE SIGNALS (negative z-score) ===")
print(value.sort_values(ascending=False).round(3))

# ============================================
# 5. RISK PARITY WEIGHTS (Inverse Vol)
# ============================================
def risk_parity_weights(returns):
    vols = returns.std() * np.sqrt(252)
    inv_vol = 1 / vols
    return inv_vol / inv_vol.sum()

returns = data.pct_change().dropna()
rp_weights = risk_parity_weights(returns)
print("\n=== RISK PARITY WEIGHTS ===")
print(rp_weights.sort_values(ascending=False).round(3))

# ============================================
# 6. COMBINE SIGNALS
# ============================================
# Normalize each signal (cross-sectional z-score)
mom_z = (momentum - momentum.mean()) / momentum.std()
val_z = (value - value.mean()) / value.std()

# Combine (equal weight to momentum and value)
combined_signal = 0.5 * mom_z + 0.5 * val_z

# Long top 4, short bottom 4 (cross-sectional)
sorted_signals = combined_signal.sort_values(ascending=False)
longs = sorted_signals.head(4)
shorts = sorted_signals.tail(4)

print("\n=== TRADE RECOMMENDATIONS ===")
print("\n🟢 LONGS (top 4):")
for ticker, sig in longs.items():
    print(f"  {ticker}: signal = {sig:+.2f}")

print("\n🔴 SHORTS (bottom 4):")
for ticker, sig in shorts.items():
    print(f"  {ticker}: signal = {sig:+.2f}")
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

# ============================================
# 1. ASSET UNIVERSE
# ============================================
universe = {
    'bonds': ['TLT', 'IEF', 'BNDX', 'IGOV'],
    'currencies': ['UUP', 'FXE', 'FXY', 'FXB'],
    'equities': ['SPY', 'EWJ', 'EFA', 'EEM'],
    'commodities': ['GLD', 'DBC', 'USO', 'CORN']
}
all_tickers = [t for sublist in universe.values() for t in sublist]

# ============================================
# 2. DOWNLOAD DATA
# ============================================
print("Downloading data...")
data = yf.download(all_tickers, period='5y', progress=False, auto_adjust=True)['Close']
data = data.dropna(how='all')
returns = data.pct_change().dropna()
print(f"Got {len(data)} days of data\n")

# ============================================
# 3. MOMENTUM SIGNAL (12-month return)
# ============================================
def momentum_signal(prices):
    return prices.pct_change(252).iloc[-1]

# ============================================
# 4. VALUE SIGNAL (3-year z-score, inverse for mean reversion)
# ============================================
def value_signal(prices):
    mean = prices.rolling(252*3).mean().iloc[-1]
    std = prices.rolling(252*3).std().iloc[-1]
    zscore = (prices.iloc[-1] - mean) / std
    return -zscore

# ============================================
# 5. CARRY SIGNAL (NEW! Add this)
# ============================================
def carry_signal():
    """
    Carry approximated by yield/income on each asset.
    Bonds: yield to maturity (proxy via 12-month dividend yield of ETF)
    Currencies: USD interest rate differential
    Equities: dividend yield
    Commodities: 0 (or futures curve slope, simplified to 0)
    """
    # Approximate yields (% per year)
    yields = {
        # Bonds (higher yield = positive carry)
        'TLT': 0.045,   # Long-term US treasuries
        'IEF': 0.040,   # 7-10 year treasuries
        'BNDX': 0.030,  # International bonds (hedged)
        'IGOV': 0.025,  # International treasuries
        
        # Currencies (USD-funded carry = foreign rate - USD rate)
        'UUP': 0.045,   # USD (Fed rate)
        'FXE': 0.025,   # EUR (ECB rate, lower than Fed)
        'FXY': 0.005,   # JPY (BoJ near-zero)
        'FXB': 0.045,   # GBP (BoE around USD)
        
        # Equities (dividend yield)
        'SPY': 0.013,
        'EWJ': 0.020,
        'EFA': 0.030,
        'EEM': 0.028,
        
        # Commodities (no yield, slight cost of carry)
        'GLD': 0.0,
        'DBC': 0.0,
        'USO': -0.05,   # Slight negative due to contango
        'CORN': -0.03
    }
    return pd.Series(yields)

# ============================================
# 6. COMPUTE ALL SIGNALS
# ============================================
momentum = momentum_signal(data)
value = value_signal(data)
carry = carry_signal()

# Z-score normalize each
mom_z = (momentum - momentum.mean()) / momentum.std()
val_z = (value - value.mean()) / value.std()
car_z = (carry - carry.mean()) / carry.std()

# Combine (Brightman-Shepherd: equal weights)
combined_signal = (mom_z + val_z + car_z) / 3

print("=== COMBINED SIGNALS (Momentum + Value + Carry) ===")
print(combined_signal.sort_values(ascending=False).round(3))

# ============================================
# 7. POSITION SIZING — Cross-Sectional Long/Short
# ============================================
# Long top 4, short bottom 4 (paper's approach: long top 1/3, short bottom 1/3)
sorted_signals = combined_signal.sort_values(ascending=False)
n_long = 4
n_short = 4

positions = pd.Series(0.0, index=combined_signal.index)
positions[sorted_signals.head(n_long).index] = 1 / n_long      # Long top 4
positions[sorted_signals.tail(n_short).index] = -1 / n_short   # Short bottom 4

print("\n=== POSITIONS ===")
print(positions[positions != 0].round(3))

# ============================================
# 8. BACKTEST (THE BIG NEW PART)
# ============================================
def backtest(data, lookback=252*3, rebalance_freq=21):
    """
    Walk-forward backtest:
    - Rebalance every 21 days (monthly)
    - Use only past data for signals (no lookahead)
    - Calculate signals at each rebalance, hold to next
    """
    returns = data.pct_change()
    
    portfolio_returns = []
    dates = []
    
    # Start after we have enough data for signals
    start_idx = lookback
    
    current_positions = pd.Series(0.0, index=data.columns)
    
    for i in range(start_idx, len(data)):
        # Rebalance every N days
        if (i - start_idx) % rebalance_freq == 0:
            # Use only past data
            past_data = data.iloc[:i]
            
            # Calculate signals
            mom = past_data.pct_change(252).iloc[-1]
            mean = past_data.rolling(252*3).mean().iloc[-1]
            std = past_data.rolling(252*3).std().iloc[-1]
            val = -(past_data.iloc[-1] - mean) / std
            car = carry_signal()
            
            # Drop NaN (assets without enough history)
            valid = mom.dropna().index.intersection(val.dropna().index).intersection(car.dropna().index)
            
            if len(valid) < 8:
                continue
            
            # Z-score
            mom_z = (mom[valid] - mom[valid].mean()) / mom[valid].std()
            val_z = (val[valid] - val[valid].mean()) / val[valid].std()
            car_z = (car[valid] - car[valid].mean()) / car[valid].std()
            
            combined = (mom_z + val_z + car_z) / 3
            sorted_sigs = combined.sort_values(ascending=False)
            
            # New positions
            current_positions = pd.Series(0.0, index=data.columns)
            n = min(4, len(sorted_sigs) // 3)
            current_positions[sorted_sigs.head(n).index] = 1 / n
            current_positions[sorted_sigs.tail(n).index] = -1 / n
        
        # Calculate today's return given current positions
        today_returns = returns.iloc[i].fillna(0)
        port_return = (current_positions * today_returns).sum()
        portfolio_returns.append(port_return)
        dates.append(data.index[i])
    
    return pd.Series(portfolio_returns, index=dates)

print("\nRunning backtest...")
strategy_returns = backtest(data)

# ============================================
# 9. PERFORMANCE STATS
# ============================================
def stats(returns, name):
    annual_ret = returns.mean() * 252
    annual_vol = returns.std() * np.sqrt(252)
    sharpe = annual_ret / annual_vol if annual_vol > 0 else 0
    cum = (1 + returns).cumprod()
    dd = (cum / cum.cummax() - 1).min()
    win_rate = (returns > 0).mean()
    
    print(f"\n=== {name} ===")
    print(f"Annual Return:    {annual_ret*100:.2f}%")
    print(f"Annual Vol:       {annual_vol*100:.2f}%")
    print(f"Sharpe Ratio:     {sharpe:.2f}")
    print(f"Max Drawdown:     {dd*100:.2f}%")
    print(f"Win Rate:         {win_rate*100:.1f}%")
    return sharpe

# Compare to SPY buy & hold
spy_returns = returns['SPY'].loc[strategy_returns.index]

s_sharpe = stats(strategy_returns, "QUANT MACRO STRATEGY")
spy_sharpe = stats(spy_returns, "SPY BUY & HOLD")

# ============================================
# 10. PLOT
# ============================================
plt.figure(figsize=(14, 7))
(1 + strategy_returns).cumprod().plot(label='Quant Macro Strategy', linewidth=2)
(1 + spy_returns).cumprod().plot(label='SPY Buy & Hold', linewidth=2)
plt.title('Systematic Global Macro vs SPY')
plt.ylabel('Cumulative Return')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

# Universe
tickers = ['SPY', 'EWJ', 'EFA', 'EEM',
           'TLT', 'IEF', 'GLD', 'DBC']

# Data
data = yf.download(tickers, period='5y', progress=False, auto_adjust=True)['Close']
data = data.dropna(how='all').ffill()
returns = data.pct_change().fillna(0)

# Simple strategy: weight by 12-month momentum, scaled by inverse vol
def simple_backtest(data, lookback=252, rebalance=21):
    rets = data.pct_change().fillna(0)
    portfolio_returns = []
    dates = []
    weights = pd.Series(0.0, index=data.columns)
    
    for i in range(lookback, len(data)):
        if (i - lookback) % rebalance == 0:
            past = data.iloc[:i]
            
            # 12-month momentum
            mom = past.pct_change(252).iloc[-1]
            # Inverse vol
            vol = past.pct_change().iloc[-60:].std() * np.sqrt(252)
            
            # Sign of momentum × inverse vol (long winners, short losers, vol-scaled)
            signals = np.sign(mom) / vol
            signals = signals.fillna(0)
            
            # Normalize so absolute weights sum to 1
            total = signals.abs().sum()
            if total > 0:
                weights = signals / total
        
        portfolio_returns.append((weights * rets.iloc[i]).sum())
        dates.append(data.index[i])
    
    return pd.Series(portfolio_returns, index=dates)

# Run
print("Running SIMPLE backtest...")
strat = simple_backtest(data)
spy = returns['SPY'].loc[strat.index]

# Stats
def stats(r, name):
    cum = (1 + r).cumprod()
    total = cum.iloc[-1] - 1
    ann_r = r.mean() * 252
    ann_v = r.std() * np.sqrt(252)
    sr = ann_r / ann_v if ann_v > 0 else 0
    dd = (cum / cum.cummax() - 1).min()
    print(f"\n=== {name} ===")
    print(f"Total: {total*100:+.2f}% | Sharpe: {sr:.2f} | Vol: {ann_v*100:.1f}% | MDD: {dd*100:.1f}%")
    return sr

s_sharpe = stats(strat, "Simple Trend + Risk Parity")
spy_sharpe = stats(spy, "SPY")

# Plot
plt.figure(figsize=(14, 7))
(1 + strat).cumprod().plot(label=f'Strategy (Sharpe {s_sharpe:.2f})', linewidth=2)
(1 + spy).cumprod().plot(label=f'SPY (Sharpe {spy_sharpe:.2f})', linewidth=2)
plt.title('Simple Quant Macro vs SPY')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
