```python

"""
MONTE CARLO OPTIONS PRICING
============================
University project scaffold — implements Monte Carlo pricing for:
  1. European call/put (compared to Black-Scholes analytical)
  2. Greeks (Delta, Gamma, Vega) via finite differences
  3. Convergence analysis (error vs simulation count)
  4. Variance reduction (antithetic variates)
  5. Asian option (path-dependent, no closed form)
  6. Barrier option (up-and-out call)

Author: Rudra Tanvar
"""
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ==========================================================
# PARAMETERS (edit these)
# ==========================================================
S0     = 100      # Current stock price
K      = 105      # Strike price
r      = 0.05     # Risk-free rate (annual)
sigma  = 0.20     # Volatility (annual)
T      = 1.0      # Time to expiry (years)
N_SIMS = 100_000  # Monte Carlo simulations
N_STEPS= 252      # Time steps per path (daily)

np.random.seed(42)

# ==========================================================
# 1. GEOMETRIC BROWNIAN MOTION PATH SIMULATOR
# ==========================================================
def simulate_gbm(S0, r, sigma, T, n_steps, n_sims, antithetic=False):
    """
    Simulate n_sims paths of geometric Brownian motion.
    dS/S = r*dt + sigma*dW  (risk-neutral measure)
    Returns array of shape (n_sims, n_steps+1).
    """
    dt = T / n_steps
    if antithetic:
        half = n_sims // 2
        Z = np.random.randn(half, n_steps)
        Z = np.vstack([Z, -Z])   # antithetic pair
    else:
        Z = np.random.randn(n_sims, n_steps)

    log_returns = (r - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * Z
    log_paths   = np.log(S0) + np.cumsum(log_returns, axis=1)
    paths       = np.exp(log_paths)
    paths       = np.hstack([np.full((n_sims,1), S0), paths])
    return paths

# ==========================================================
# 2. BLACK-SCHOLES ANALYTICAL (for comparison)
# ==========================================================
def bs_call(S0, K, r, sigma, T):
    d1 = (np.log(S0/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    return S0 * norm.cdf(d1) - K * np.exp(-r*T) * norm.cdf(d2)

def bs_put(S0, K, r, sigma, T):
    d1 = (np.log(S0/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    return K * np.exp(-r*T) * norm.cdf(-d2) - S0 * norm.cdf(-d1)

# ==========================================================
# 3. EUROPEAN CALL/PUT VIA MONTE CARLO
# ==========================================================
def mc_european(S0, K, r, sigma, T, n_sims, n_steps, option='call',
                antithetic=False):
    paths = simulate_gbm(S0, r, sigma, T, n_steps, n_sims, antithetic)
    ST = paths[:, -1]
    if option == 'call':
        payoffs = np.maximum(ST - K, 0)
    else:
        payoffs = np.maximum(K - ST, 0)
    price = np.exp(-r*T) * payoffs.mean()
    std_err = np.exp(-r*T) * payoffs.std() / np.sqrt(n_sims)
    return price, std_err

# ==========================================================
# 4. GREEKS VIA FINITE DIFFERENCES
# ==========================================================
def compute_greeks(S0, K, r, sigma, T, n_sims, n_steps, h=1.0):
    """
    Delta = dV/dS   (bump S up/down, revalue)
    Gamma = d²V/dS²
    Vega  = dV/dσ
    """
    # Delta and Gamma
    up,   _ = mc_european(S0+h, K, r, sigma, T, n_sims, n_steps)
    base, _ = mc_european(S0,   K, r, sigma, T, n_sims, n_steps)
    down, _ = mc_european(S0-h, K, r, sigma, T, n_sims, n_steps)
    delta = (up - down) / (2*h)
    gamma = (up - 2*base + down) / (h**2)

    # Vega
    dsig = 0.01
    vega_up,   _ = mc_european(S0, K, r, sigma+dsig, T, n_sims, n_steps)
    vega_down, _ = mc_european(S0, K, r, sigma-dsig, T, n_sims, n_steps)
    vega = (vega_up - vega_down) / (2*dsig)

    return delta, gamma, vega

# ==========================================================
# 5. ASIAN OPTION (arithmetic average — no closed form)
# ==========================================================
def mc_asian_call(S0, K, r, sigma, T, n_sims, n_steps):
    paths = simulate_gbm(S0, r, sigma, T, n_steps, n_sims)
    avg   = paths[:, 1:].mean(axis=1)     # average price over life
    payoffs = np.maximum(avg - K, 0)
    return np.exp(-r*T) * payoffs.mean()

# ==========================================================
# 6. BARRIER OPTION (up-and-out call)
# ==========================================================
def mc_up_and_out_call(S0, K, B, r, sigma, T, n_sims, n_steps):
    """If price hits B any time during life -> option knocks out (worth 0)"""
    paths = simulate_gbm(S0, r, sigma, T, n_steps, n_sims)
    max_price = paths.max(axis=1)
    knocked_out = max_price >= B
    ST = paths[:, -1]
    payoffs = np.where(knocked_out, 0, np.maximum(ST - K, 0))
    return np.exp(-r*T) * payoffs.mean()

# ==========================================================
# RUN ANALYSIS
# ==========================================================
print("="*60)
print("MONTE CARLO OPTIONS PRICING")
print(f"S0={S0}  K={K}  r={r}  σ={sigma}  T={T}  Sims={N_SIMS:,}")
print("="*60)

# --- European Call
mc_c, se_c = mc_european(S0, K, r, sigma, T, N_SIMS, N_STEPS, 'call')
bs_c = bs_call(S0, K, r, sigma, T)
print(f"\nEUROPEAN CALL")
print(f"  Monte Carlo:    ${mc_c:.4f}  (±${1.96*se_c:.4f} 95% CI)")
print(f"  Black-Scholes:  ${bs_c:.4f}")
print(f"  MC Error:       {abs(mc_c-bs_c)/bs_c*100:.3f}%")

# --- European Put
mc_p, se_p = mc_european(S0, K, r, sigma, T, N_SIMS, N_STEPS, 'put')
bs_p = bs_put(S0, K, r, sigma, T)
print(f"\nEUROPEAN PUT")
print(f"  Monte Carlo:    ${mc_p:.4f}")
print(f"  Black-Scholes:  ${bs_p:.4f}")

# --- With antithetic variates
mc_c_av, se_c_av = mc_european(S0, K, r, sigma, T, N_SIMS, N_STEPS,
                                'call', antithetic=True)
print(f"\nEUROPEAN CALL — antithetic variates")
print(f"  Price:          ${mc_c_av:.4f}")
print(f"  Std error:      ${se_c_av:.4f}  ({(1-se_c_av/se_c)*100:.1f}% reduction)")

# --- Greeks
print(f"\nGREEKS (finite difference)")
delta, gamma, vega = compute_greeks(S0, K, r, sigma, T, 50_000, N_STEPS)
print(f"  Delta: {delta:.4f}")
print(f"  Gamma: {gamma:.4f}")
print(f"  Vega:  {vega:.4f}")

# --- Asian
asian = mc_asian_call(S0, K, r, sigma, T, N_SIMS, N_STEPS)
print(f"\nASIAN CALL (arithmetic avg): ${asian:.4f}")
print(f"  (Cheaper than European because averaging reduces volatility)")

# --- Barrier
barrier = 120
up_out = mc_up_and_out_call(S0, K, barrier, r, sigma, T, N_SIMS, N_STEPS)
print(f"\nUP-AND-OUT CALL (barrier=${barrier}): ${up_out:.4f}")
print(f"  (Cheaper than European because knock-out risk)")

# ==========================================================
# CONVERGENCE STUDY
# ==========================================================
print(f"\n{'='*60}")
print(f"CONVERGENCE STUDY")
print(f"{'='*60}")
sim_counts = [1_000, 5_000, 10_000, 50_000, 100_000, 500_000]
mc_prices, se_list = [], []
for n in sim_counts:
    p, se = mc_european(S0, K, r, sigma, T, n, N_STEPS, 'call')
    mc_prices.append(p)
    se_list.append(se)
    print(f"  N={n:>7,}  Price=${p:.4f}  StdErr=${se:.4f}")
```
